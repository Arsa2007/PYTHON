```python
import pandas as pd
from typing import List,Union
list1=[[10,'ab'],[12,'bc'],[14,'de']]
DF=pd.DataFrame(list1, columns=['age','name'])



def cc(list1:List[List[Union[int,str]]])->pd.DataFrame:

   Dframe=pd.DataFrame(list1, columns=['age','name'])
   #for number of rows and columns
   RC=Dframe.shape
   print(RC)

   #for number of rows
   R1=len(Dframe.index)
   print(f'Rows={R1}')

   #for number of columns
   C1=len(Dframe.columns)
   print(f'Columns={C1}')

   return pd.DataFrame(list1, columns=['age','name'])
cc(list1)

def getDataframeSize(DF:pd.DataFrame) -> List:
    return [DF.shape[0], DF.shape[1]]
getDataframeSize(DF)
```
