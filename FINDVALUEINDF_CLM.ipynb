```python
import pandas as pd
list1=[[101,"abc",19],[102,"efg",18],[104,"def",20]]
students=pd.DataFrame(list1,columns=['student_id','name','age'])

def selectData(students: pd.DataFrame) -> pd.DataFrame:
  x=int(input("Enter student id: "))
   #students['student_id]'go to students and under the column student_id
   #where condition that is student['student_id']==x
  return students['student_id'].where(students['student_id']==x)
selectData(students)
```
