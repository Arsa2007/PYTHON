#  CNN MODEL 

```python
#CNN


from tensorflow.keras.layers import Input,Conv2D,MaxPooling2D,Dense,Flatten
from tensorflow.keras.models import Sequential,Model
model_1=Sequential([
    
    Conv2D(512,1,use_bias=True,activation='relu',input_shape=(512,512,3)),
    #512 filters, 3 =kernel, 2=strides
    #512*3*3+512*1
    MaxPooling2D(pool_size=8),
    
    Conv2D(512,3,use_bias=True,activation='relu'),
    MaxPooling2D(pool_size=4),

    Conv2D(256,3,use_bias=True,activation='relu'),
    MaxPooling2D(pool_size=2),
    
    Conv2D(128,3,use_bias=True,activation='relu'),
    MaxPooling2D(pool_size=1),
    
    Flatten(),
    #activation function softplus and relu use
    Dense(64,use_bias=True,activation='relu'),
    Dense(3,activation='softmax')])

```

```python
model_1.summary()
model_1.save()
#compiled the model 
optimizer = tf.keras.optimizers.Adam(learning_rate=0.001)
model_1.compile(optimizer=optimizer,loss=tf.keras.losses.SparseCategoricalCrossentropy(),metrics=['accuracy'])
```

```python
import cv2
import imghdr
import os
#creating a data directory 
data_dir='D:\Test\TEST'
#list ,types of format to accept
img_ex=['jpeg','jpg','png']
 

#to check for correct format
#try and except block to check and remove if not correct format

for image_class in os.listdir(data_dir):
    for image in os.listdir(os.path.join(data_dir,image_class)):
        image_path=os.path.join(data_dir,image_class,image)
        try:
            img=cv2.imread(image_path)
            tip=imghdr.what(image_path)
            if tip not in img_ex :
                print(f'Image not exist in list {image}'.format(image_path))
                os.remove(image_path)
        except Exception as e:
            print(f'Image has issue {image}'.format(image_path))

            
```

##  CREATING DATAPIPELING

```python
import numpy
import tensorflow as tf
#creating pipeline 
data=tf.keras.utils.image_dataset_from_directory(data_dir,color_mode='rgb',batch_size=2,image_size=(512,512))
#convetor the image as numpyarray to iterate, so that it can be loaded
data_iterator=data.as_numpy_iterator()
#to iterate, for loops
data_iterator.next()
#scaled version to covert the vector between 0 and 1
data=data.map(lambda x,y:(x/255,y))
scaled_iterator=data.as_numpy_iterator()
batch=scaled_iterator.next()
```

```python
#creating data sets according foor the model to train, validate and test
train_size=int(len(data)*.7)
val_size=int(len(data)*.2)
test_size=int(len(data)*.1)+1
#after we det the len of data we use skip and take ti intake data for the model
train=data.take(train_size)
val=data.take(val_size).skip(train_size)
test=data.take(test_size).skip(train_size+val_size)
```

```python
import tkinter as tk
from tkinter import filedialog, messagebox
from PIL import Image, ImageTk
import numpy as np
import os

# Try to import tensorflow and load the model
model = None
model_path = 'M1.h5'  
IMG_SIZE = (512, 512)     
class_names = ['Benign', 'Malignant', 'Normal']  

try:
    import tensorflow as tf
    if os.path.exists(model_path):
        model = tf.keras.models.load_model(model_path)
    else:
        print(f"Model file '{model_path}' not found.")
except ImportError:
    print("TensorFlow is not installed. Please install it with 'pip install tensorflow'.")

def select_and_predict():
    if model is None:
        messagebox.showerror("Model Not Found", f"Model file '{model_path}' not found or TensorFlow not installed.")
        return

    file_path = filedialog.askopenfilename(
        filetypes=[("Image Files", "*.png;*.jpg;*.jpeg;*.bmp;*.gif")]
    )
    if file_path:
        # Display the image in Tkinter
        img = Image.open(file_path)
        img_display = img.resize((512, 512))
        tk_img = ImageTk.PhotoImage(img_display)
        image_label.config(image=tk_img)
        image_label.image = tk_img

        # Preprocess for CNN
        img_model = img.resize(IMG_SIZE)
        img_array = np.array(img_model)
        # If image is grayscale, convert to RGB
        if len(img_array.shape) == 2:
            img_array = np.stack((img_array,)*3, axis=-1)
        # If image has alpha channel, remove it
        if img_array.shape[-1] == 4:
            img_array = img_array[..., :3]
        img_array = img_array.astype('float32') / 255.0  # Normalize
        img_array = np.expand_dims(img_array, axis=0)    # Add batch dimension

        # Predict
        prediction = model.predict(img_array)
        predicted_class = class_names[np.argmax(prediction)]
        confidence = np.max(prediction) * 100
        result_label.config(text=f"Prediction: {predicted_class} ({confidence:.2f}%)")

root = tk.Tk()
root.title("CNN Predictor")

tk.Button(root, text="Select Image", command=select_and_predict).pack(pady=10)
image_label = tk.Label(root)
image_label.pack(pady=10)
result_label = tk.Label(root, text="", font=("Arial", 14))
result_label.pack(pady=10)

root.mainloop()
```
