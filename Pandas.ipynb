```python

#SERIES FUNCTION
#one dimensional
import numpy as np
import pandas as pd


data=[1,2,3,4,5]
s=pd.Series(data)
print(s)

s=pd.Series(data,index=['a','b','c','d','e'])
print(s)
```

```python
#dataframe
data={
    'Name':['Alice','Bob','Alan'],
    'Age':[16,15,17],
    'City':['New York','London','Paris']
}

df=pd.DataFrame(data)
print(df)
#to save data in csv file
df.to_csv('data.csv',index=False)
```
