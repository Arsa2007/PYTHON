```python
import os
def create(filename):
 try:
  with open(filename, 'x') as f:
    print(f'{filename} Created succesfully')
 except FileNotFoundError:
  print(f"{filename} doesn't exist")
 except Exception as e :
  print('An error occured')

def opened(filename):
 try:
  with open(filename, 'r') as f:
    print(f.read())
    print(f'{filename} Open succesfully')
 except FileNotFoundError:
  print(f"{filename} doesn't exist")
 except Exception as e :
  print('An error occured')

def removes(filename):
  try:
    os.remove(filename)
    print(f'{filename} Removed succesfully')
  except FileNotFoundError:
    print(f"{filename} doesn't exist")
  except Exception as e :
    print('An error occured')

def update(filename):
  try:
    with open(filename,'a') as f:
     content=input('Enter your data')
     f.write(content)
     print(f'{filename} Updated succesfully')
  except FileNotFoundError:
    print(f"{filename} doesn't exist")
  except Exception as e :
    print('An error occured')

def list_dir():
  dir=os.listdir()
  if len(dir)==0:
    print('No files')
  else:
    for i in dir:
      print(i)


def main():

 while True:
  num=int(input('Enter your choice'))
  if num==1:
   filename=input('Enter your filename:')
   create(filename)
  elif num==2:
   filename=input('Enter your filename:')
   opened(filename)
  elif num==3:
   filename=input('Enter your filename:')
   removes(filename)
  elif num==4:
   filename=input('Enter your filename:')
   update(filename)
  elif num==5:
   list_dir()

print(main())
```
