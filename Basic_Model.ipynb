```python
import numpy as np
import tensorflow as tf
from tensorflow import keras
model=keras.Sequential([tf.keras.Input(shape=(1,)),tf.keras.layers.Dense(units=1)])


model.compile(optimizer='sgd',loss='mean_squared_error')
xs=np.array([1.2,1.3,1.4,1.5,1.5])
ys=np.array([1.0,1.1,1.2,1.3,1.3])
model.fit(xs,ys,epochs=10)

model.predict(np.array([1.0]))
```
