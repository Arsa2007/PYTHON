```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import re
import string
import nltk
from nltk.corpus import stopwords
nltk.download('stopwords')

```

```python
with open('/content/Restaurant_Reviews.tsv','r') as tsv:
 with open('/content/Reviews.csv','w') as csv:
  for i in tsv:
    content=re.sub("\t", ",",i)
    csv.write(content)
```

```python
tw=pd.read_csv("output.csv")
tw[[' Review']].head()



```

```python
def clean( Review):
  Review=Review.lower()
  Review=re.sub('https?://\S+|www\.\S+','',Review)
  Review =Review.translate(str.maketrans('', '', string.punctuation))
  Review = ''.join([i for i in Review if not i.isdigit()])
  words = Review.split()
  words = [w for w in words if w not in stopwords.words('english')]
  return ' '.join(words)

tw['Cleaned_reviews']=tw[' Review'].apply(clean)
tw.head()
```
