```python
import pandas as pd
list1=[[101,"abc",19],[102,"efg",18],[104,"def",20]]
students=pd.DataFrame(list1,columns=['id','name','age'])

def selectData(students: pd.DataFrame) -> pd.DataFrame:
    return students.loc[students['id']==101,['name','age']]
selectData(students)
```
