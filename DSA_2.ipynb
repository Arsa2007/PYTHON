```python
#diagonal of a matrix
list1=[]
diagonal=[]

n=int(input())

for i in range(n):
    list1.append([int(ele) for ele in input().split()])

    
for i in range(n):
            diagonal.append(list1[i][i])
    
print(diagonal) 
print(list1)
```
